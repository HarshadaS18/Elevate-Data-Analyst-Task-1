```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
#Check 4: Check unwanted spaces
for col in df.select_dtypes(include='object').columns:
    print(col, df[col].astype(str).str.strip().ne(df[col].astype(str)).sum())
```

```python
# Understand the warning
for col in df.select_dtypes(include='str').columns:
    print(
        col,
        df[col].notna().mul(
            df[col].str.strip().ne(df[col]).fillna(False)
        ).sum()
    )
```

```python
df.loc[
    df['title'].notna() &
    df['title'].str.strip().ne(df['title']),
    'title'
]
```

```python
# Remove unwanted spaces
for col in df.select_dtypes(include='str').columns:
    df[col] = df[col].str.strip()
```

```python
#Verify again
for col in df.select_dtypes(include='str').columns:
    count = (
        df[col].notna() &
        df[col].str.strip().ne(df[col]).fillna(False)
    ).sum()

    print(col, count)
```

```python
#Check 5: Check inconsistent text

print(df['type'].value_counts(dropna=False))
```

```python
#Check 6: Check invalid values

# Check release years
print(df['release_year'].describe())

# Check unexpected ratings
print(df['rating'].value_counts())

# Check duration
print(df['duration'].value_counts())
```

```python
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nUnique Content Types:")
print(df['type'].value_counts())

print("\nExtra Spaces:")
for col in df.select_dtypes(include='str').columns:
    count = (
        df[col].notna() &
        df[col].str.strip().ne(df[col]).fillna(False)
    ).sum()
    print(col, count)
```

```python
df.to_csv("Netflix_Cleaned.csv", index=False)

df.to_excel("Netflix_Cleaned.xlsx", index=False)
```

```python
# Identify rows where rating contains minute values (e.g., 'min')
invalid_ratings = df['rating'].str.contains('min', na=False)
print(f"Rows with misplaced ratings: {invalid_ratings.sum()}")

# View an example of these rows
print(df[invalid_ratings][['title', 'rating', 'duration']])
```

```python
df['director'] = df['director'].fillna('Unknown Director')
df['cast'] = df['cast'].fillna('Unknown Cast')
df['country'] = df['country'].fillna('Unknown Country')
df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df['type'] = df['type'].str.strip()
df['title'] = df['title'].str.strip()
```

```python
df.to_csv('netflix_cleaned_dataset.csv', index=False)
print("Cleaned dataset successfully saved!")
```
